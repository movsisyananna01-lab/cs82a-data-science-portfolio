```python
import pandas as pd

df = pd.read_csv("sales_clean.csv")
print(df.shape) 
```

```python
print("Price mean:", df["price"].mean())
print("Price median:", df["price"].median())
print("Price standard deviation:", df["price"].std())

print("Qty mean:", df["qty"].mean())
print("Qty median:", df["qty"].median())
print("Qty standard deviation:", df["qty"].std())
```

I would report the median of $37.53 to show the typical price. A few high prices raise the mean to 55.52. 

```python
df["price"].plot(kind="hist", bins=20) 
```

```python
df["revenue"] = df["price"] * df["qty"] 
df.head() 
```

```python
df.corr(numeric_only=True) 
```

```python
df.plot.scatter(x="price", y="revenue") 
df.plot.scatter(x="qty", y="revenue");
```

Higher prices tend to go with higher revenue, though a few sales with unusuallyhigh revene could affect the strength of this relationship.
