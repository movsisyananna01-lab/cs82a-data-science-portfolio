```python
import pandas as pd 

books = pd.read_csv("allbacks.csv") 
books.head()
```

```python
covers = sorted(books["cover"]) 
print("Median cover:", covers[7]) 
print("Mean weight of all books:", books["weight"].mean())
print("Mean weight by cover:") 
print(books.groupby("cover")["weight"].mean())
```

The median cover type is pb. 
The average weight of all 15 books is about 707g. 
Hb weigh about 796g, pb weigh 628g. The difference is about 168g.
